# Dusra Tarika State (Structure Banane kaa)🌌

Pydantic check karta hai ki jo data hum de rahe hain, woh expected structure aur rules ke according hai ya nahi.
BaseModel Pydantic ka main class hai.
State ek Pydantic model hai.
Decorator Yaani jab score ki value model mein enter hogi, validator usko check karega.


```python
'''


User Data
   ↓
State()
   ↓
Pydantic BaseModel
   ↓
topic → str
score → int
summary → str
   ↓
score validator
   ↓
Is score < 0?
   │
   ├── YES → ValueError 
   │
   └── NO  → return value 


'''
```

```python
from pydantic import BaseModel, field_validator


class State(BaseModel):
    topic: str
    score: int
    summary: str = ""

    @field_validator("score")
    @classmethod
    def score_positive(cls, v):
        if v < 0:
            raise ValueError("Score Must Be Positive")
        return v
```
